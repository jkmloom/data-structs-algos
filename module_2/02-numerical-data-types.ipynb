# numerical data types

```python
x = 5 # integer
print(x)
price = 26.25 # floating point
print(price)
number  = 3e6 # where 3e6 = 3 * 10^6 = 3,000,000
print(number)
```

```python
# complex numbers
# example >> 4 + 5i in python will become 4 + 5j
z = 4 + 5j
print(z)
```

```python
# find/check variable type
print(type(z))
```

```python
print(type(x))
print(type(price))
print(type(number))
```

```python
# abs() method
print(abs(-5))
```

```python
# pow() method
print(pow(2, 3)) # 2^3
```

```python
# round() method
print(round(4.21))
print(round(4.50))
print(round(4.75))
```

```python
# min() and max() mehtods
nums = [7, 3, 5, 8, 2, 9, 5, 6]
print(min(nums)) # returns the minimum value
print(max(nums)) # returns the maximum value
```

```python
# sum() method
print(sum(nums))
```
